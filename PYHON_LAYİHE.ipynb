## 6.Müxtəlif qruplar üçün orta məmnunluq səviyyələrini hesablayın (məsələn, şöbə, cins, təhsil səviyyəsi üzrə).

```python
performance_df=pd.read_csv(r"D:\INDIRILENLER\MATRIX\PROJECT\1. HR Analytics Case Study DA project\HR Analytics Case Study\Performance Rating.csv")
```

```python
merged_df_emp_perform = pd.merge(employee_df, performance_df, on='EmployeeID')
```

```python
merged_df_emp_ed_perform=pd.merge(merged_df_emp_ed,performance_df,on='EmployeeID')
```

```python
# Şöbə üzrə orta məmnunluq səviyyəsi
avg_satisfaction_by_dept = merged_df_emp_perform.groupby("Department")[["EnvironmentSatisfaction", "JobSatisfaction", "RelationshipSatisfaction","WorkLifeBalance"]].mean()
print("Şöbə üzrə orta məmnunluq səviyyəsi:\n", avg_satisfaction_by_dept)

print()

# Cins üzrə orta məmnunluq səviyyəsi
avg_satisfaction_by_gender = merged_df_emp_perform.groupby("Gender")[["EnvironmentSatisfaction", "JobSatisfaction", "RelationshipSatisfaction","WorkLifeBalance"]].mean()
print("\nCins üzrə orta məmnunluq səviyyəsi:\n", avg_satisfaction_by_gender)

print()

# Təhsil səviyyəsi üzrə orta məmnunluq səviyyəsi
avg_satisfaction_by_ed = merged_df_emp_ed_perform.groupby("EducationLevel")[["EnvironmentSatisfaction", "JobSatisfaction", "RelationshipSatisfaction","WorkLifeBalance"]].mean()
print("\nTəhsil səviyyəsi üzrə orta məmnunluq səviyyəsi:\n", avg_satisfaction_by_ed)
```

## 7.	Məmnunluq səviyyələri ilə attrition arasındakı əlaqəni araşdırın.

```python
merged_df_emp_perform['AttritionFlag'] = merged_df_emp_perform['Attrition'].map({'Yes': 1, 'No': 0})

attrition_by_job_sat = merged_df_emp_perform.groupby('JobSatisfaction')['AttritionFlag'].mean()
print(attrition_by_job_sat)

print()

attrition_by_envr_sat = merged_df_emp_perform.groupby('EnvironmentSatisfaction')['AttritionFlag'].mean()
print(attrition_by_envr_sat)

print()

attrition_by_relationship_sat = merged_df_emp_perform.groupby('RelationshipSatisfaction')['AttritionFlag'].mean()
print(attrition_by_relationship_sat)

print()

attrition_by_WorkLifeBalance_sat = merged_df_emp_perform.groupby('WorkLifeBalance')['AttritionFlag'].mean()
print(attrition_by_WorkLifeBalance_sat)
```

```python
#məmnunluq səviyyələri (işdən məmnunluq, ətraf mühit, münasibətlər və iş-həyat balansı) ilə attrition (işdən ayrılma) arasında 
#gözlənildiyi qədər güclü və birbaşa əlaqə müşahidə olunmur.
#Bəzi hallarda məmnunluq səviyyəsi artdıqca attrition-un da artdığı görünür. 
#Bu isə onu göstərir ki, işdən ayrılma qərarına yalnız məmnunluq səviyyələri təsir etmir.
#Bununla yanaşı, maaş, vəzifə, inkişaf imkanları, idarəetmə tərzi və şəxsi səbəblər kimi digər amillər də əhəmiyyətli rol oynaya bilər.
```

## 8.	SelfRating və ManagerRating-in paylanmasını yoxlayın.

## 12.Departamentin fəaliyyətini, iş məmnunluğunu və işdən çıxma dərəcələrini qiymətləndirin.

```python
#Fəaliyyət dedikdə orta məmnunluq yüksək, attrition dərəcəsi aşağı olan şöbələr müsbət qiymətləndirilir.
#Yüksək məmnunluq → yaxşı idarəetmə, uyğun mühit.
#Aşağı attrition → işçilər iş yerindən razıdır, uzunmüddətli qalırlar.
# Bu nəticələrə əsasən, Sales departamentində ən yüksək işdən çıxma dərəcəsi var, 
# Technology departamentində isə ən aşağısıdır. 
# Həmçinin, iş məmnunluğu səviyyəsi demək olar ki, 
# hər departamentdə bənzərdir (3.43-3.44 arası), amma ümumi məmnunluq hələ də yüksək deyil.
# Sales departamentində işçilərin işdən çıxma ehtimalının nisbətən yüksək olduğu görünür.
```

```python
#1.Departament üzrə iş məmnunluğunun ortalamasını tapmaq
avg_job_satisfaction_by_dept = merged_df_emp_perform.groupby("Department")["JobSatisfaction"].mean()
print(avg_job_satisfaction_by_dept)

print()

# 2.Departament üzrə işdən çıxma dərəcəsini hesablamaq (attrition)
attrition_rate_by_dept = merged_df_emp_perform.groupby("Department")["AttritionFlag"].mean()
print(attrition_rate_by_dept)
```

## 13.Yüksək nəticə göstərən və aşağı nəticə göstərən şöbələri müəyyən edin.

## 14.Təşkilat daxilində gender bölgüsü, maaş fərqləri və müxtəlifliyi təhlil edin

```python
#Gender üzrə maaş fərqləri
avg_salary_by_gender=employee_df.groupby("Gender")["Salary"].mean()
print(avg_salary_by_gender)

print()

median_salary_by_gender = employee_df.groupby('Gender')['Salary'].median()
print(median_salary_by_gender)


# Mean (Orta) maaşda "Prefer Not To Say" olan qrupun maaşı xeyli yüksək görünür — bu onu göstərir ki, 
# həmin qrupda çox yüksək maaş alan biri və ya bir neçəsi var. Bu, ortanı süni şəkildə qaldıra bilər.

# Median (Ortancıl) maaş isə bu yüksək maaşların təsirindən kənarda qalır və daha real görünür.
```

```python
#Etnik müxtəliflik 
ethnicity_distribution = employee_df['Ethnicity'].value_counts()
print(ethnicity_distribution)

print()

ethnicity_distribution_percent = employee_df['Ethnicity'].value_counts(normalize=True) * 100
print(ethnicity_distribution_percent)

sns.countplot(data=employee_df, y='Ethnicity', order=employee_df['Ethnicity'].value_counts().index)
plt.title('Ethnic Diversity in the Organization')
plt.xlabel('Number of Employees')
plt.ylabel('Ethnicity')
plt.show()
```
